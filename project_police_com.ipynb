```python
#connect to mysql
conn = pymysql.connect(
    host="localhost",
    user="root",
    password="root",
)
cursor = conn.cursor()
cursor.execute("create database police_log")
```

```python
cursor.execute("USE police_log;")
```

```python
cursor.execute("""CREATE TABLE IF NOT EXISTS police_check (
               id INT AUTO_INCREMENT PRIMARY KEY,
               stop_date date,
               stop_time time,
               country_name varchar(100),
               driver_gender varchar(10),
               driver_age int,
               driver_race varchar(50),
               violation_raw varchar(100),
               violation varchar(100),
               search_conducted boolean,
               search_type varchar(100),
               stop_outcome varchar(100),
               is_arrested boolean,
               stop_duration varchar(50),
               drugs_related_stop boolean,
               vehicle_number varchar(50)
)""")    
```

```python
#prepare insert query
insert_query = """
INSERT INTO police_check (
stop_date, stop_time, country_name,	driver_gender, driver_age,	driver_race, violation_raw,	violation, search_conducted, search_type, stop_outcome,
is_arrested, stop_duration,	drugs_related_stop,	vehicle_number) VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s,%s)
"""
```

```python
#Using iterrows to insert Rows
for index, row in police_ledger.iterrows():
    cursor.execute(insert_query, ( 
    row['stop_date'],
    row['stop_time'],
    row['country_name'],
    row['driver_gender'],
    row['driver_age'],
    row['driver_race'],
    row['violation_raw'],
    row['violation'],
    row['search_conducted'],
    row['search_type'],
    row['stop_outcome'],
    row['is_arrested'],
    row['stop_duration'],
    row['drugs_related_stop'],
    row['vehicle_number']
))    
conn.commit()
cursor.close()
conn.close()
print("Data inserted successfully using iterrows()!")
```
